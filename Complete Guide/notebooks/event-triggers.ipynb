# GitHub Actions Event Triggers

## `on` keyword is used to define event triggers

### Push & Pull Request Events
- **`push`** – Triggers on push to a branch or tag
- **`pull_request`** – Triggers on pull request activity
- **`pull_request_target`** – Triggers on pull request from a fork

### Scheduled Events
- **`schedule`** – Triggers on a cron schedule
    ```yaml
    on:
        schedule:
            - cron: '0 0 * * *'
    ```

### Manual Events
- **`workflow_dispatch`** – Manually triggered from the GitHub UI or API
- **`repository_dispatch`** – Triggered via GitHub REST API

### Workflow Events
- **`workflow_call`** – Triggered when called from another workflow
- **`workflow_run`** – Triggers when another workflow runs

### Repository Events
- **`create`** – Branch or tag created
- **`delete`** – Branch or tag deleted
- **`fork`** – Repository forked
- **`issues`** – Issue activity
- **`issue_comment`** – Comment on an issue or PR
- **`label`** – Label activity
- **`milestone`** – Milestone activity
- **`release`** – Release activity
- **`star`** – Repository starred
- **`watch`** – Repository watched

### Code & Security Events
- **`check_run`** – Check run activity
- **`check_suite`** – Check suite activity
- **`deployment`** – Deployment created
- **`deployment_status`** – Deployment status update
- **`page_build`** – GitHub Pages build
- **`status`** – Commit status update
- **`security_advisory`** – Security advisory published

### Collaboration Events
- **`member`** – Collaborator added
- **`project`** – Project board activity
- **`project_card`** – Project card activity
- **`project_column`** – Project column activity
- **`public`** – Repository made public
- **`registry_package`** – Package published or updated
- **`gollum`** – Wiki page updated
- **`discussion`** – Discussion activity
- **`discussion_comment`** – Discussion comment activity
